### Import Libraries

```python
from pathlib import Path

import kagglehub
import pandas as pd
```

### Download the Dataset
KaggleHub returns the folder containing the downloaded files. If you have already downloaded `IMDB Dataset.csv` from the source link, place it beside this notebook; the code will use that local copy instead.

```python
dataset_id = "lakshmi25npathi/imdb-dataset-of-50k-movie-reviews"
local_csv = Path("IMDB Dataset.csv")

if local_csv.is_file():
    dataset_folder = local_csv.resolve().parent
    print("Using the locally downloaded dataset.")
else:
    dataset_folder = Path(kagglehub.dataset_download(dataset_id))

print("Dataset folder:", dataset_folder)
```

### Locate the CSV File

```python
csv_path = dataset_folder / "IMDB Dataset.csv"

if not csv_path.is_file():
    raise FileNotFoundError(f"The expected dataset file was not found: {csv_path}")

print("CSV file:", csv_path.name)
```

### Open the Dataset
Each row represents one movie review. `head()` displays the first five rows so we can inspect the text and its label.

```python
df = pd.read_csv(csv_path)

required_columns = {"review", "sentiment"}
if not required_columns.issubset(df.columns):
    raise ValueError("The dataset must contain 'review' and 'sentiment' columns.")

print(f"Number of reviews: {df.shape[0]:,}")
print(f"Number of columns: {df.shape[1]}")
df.head()
```

## Explore the Dataset

### Dataset Information
`info()` shows the column names, data types, non-null counts, and memory usage.

```python
df.info()
```

### Summary Statistics
`describe(include="all")` summarizes the text columns, including the number of entries and unique values. It does not measure review length.

```python
df.describe(include="all")
```

### Class Distribution
Compare the number and percentage of positive and negative reviews to see whether the classes are balanced.

```python
sentiment_counts = df["sentiment"].value_counts()
sentiment_percentages = df["sentiment"].value_counts(normalize=True).mul(100)

class_distribution = pd.DataFrame({
    "Number of reviews": sentiment_counts,
    "Percentage (%)": sentiment_percentages.round(2),
})
class_distribution
```

### Missing Values
Missing review text or sentiment labels could affect later analysis. This check reports missing values without modifying the original data.

```python
df.isna().sum().rename("Missing values").to_frame()
```

### Duplicate Rows
Repeated review-label pairs can affect later training and evaluation. We count them here; deciding whether to remove them belongs to a later preprocessing step.

```python
duplicate_count = df.duplicated().sum()
print(f"Duplicate rows: {duplicate_count:,}")
```
