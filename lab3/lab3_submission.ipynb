## Unigram

**Unigram** is an n-gram consisting of a single item of a sequence.

- It is commonly used to calculate the probability of finding a word in an article.

$$
P(A) = \frac{\text{count of } A}{\text{count of all words}}
$$

```python
import nltk
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

string = "moses supposes his toeses are roses but moses supposes erroneously"

unigrams = ngrams(word_tokenize(string), 1)

for item in unigrams:
    print(item)
```

### Calculating probabilities in unigram models

```python
from collections import Counter

# Count complete tokens instead of matching substrings.
word_counts = Counter(word_tokenize(string))
words_frq = sorted(word_counts.items())
words_frq
```

```python
### Calculate probabilities:

probab = []

for count in range(len(words_frq)):
    pro = words_frq[count][1]/len(word_tokenize(string))
    probab.append((words_frq[count][0],pro))

print('The probabilities are: ', probab)
```

## Bigram

**Bigram** an n-gram consisting of two items of a sequence
- Two words that are used together to mean something specific

$$
P(B \mid A) = \frac{\text{Count of }(AB)}{\text{Count of }(A)}
$$

### Method 1

```python
import nltk
from nltk.tokenize import word_tokenize

string = "moses supposes his toeses are roses but moses supposes erroneously"

bigrams = nltk.bigrams(word_tokenize(string))
for i in bigrams:
    print(i)
```

### Method 2

```python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

string = "moses supposes his toeses are roses but moses supposes erroneously"

bigrams = ngrams(word_tokenize(string),2)

for i in bigrams:
    print(i)
```

## What is padding in NLP?

Padding is a technique used to ensure that all input sequences have the same length. This is necessary because neural networks require inputs that have the same shape and size. However, when we pre-process and use texts as inputs for our model, not all sentences have the same length. In other words, some sentences are longer or shorter than others. We need to have inputs with the same size, and this is where padding comes in.

Padding is a special form of masking where the masked steps are at the start or the end of a sequence. Padding comes from the need to encode sequence data into contiguous batches: in order to make all sequences in a batch fit a given standard length.
The padding is added before splitting the sentence.

**n-1 padding is added.**



```python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize


string = "moses supposes his toeses are roses but moses supposes erroneously"

print('------------------- Padding in Bigrams ------------------------------')
list(ngrams(word_tokenize(string), pad_left=True, left_pad_symbol="<s>",
                                pad_right=True, right_pad_symbol="</s>", n=2))
```

```python
print('------------------- Padding in Trigrams ------------------------------')
list(ngrams(word_tokenize(string), pad_left=True, left_pad_symbol="<s>",
                                pad_right=True, right_pad_symbol="</s>", n=3))
```

Alternatively, We can get rid of the parameters by using
> pad_both_ends method in NLTK

**Note the n argument, that tells the function we need padding for bigrams.**

```python
from nltk.lm.preprocessing import pad_both_ends
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

string = "moses supposes his toeses are roses but moses supposes erroneously"

list(ngrams(pad_both_ends(word_tokenize(string), n=2), n=2))
```

### Training the Maximum Likelihood Model

Maximum likelihood estimation estimates the model parameters such that the probability is maximized.

- Get the requisite n_grams frequency counts from a corpus.

- Normalize them to a 0-1 range.


- Build the vocabulary (of unique words): to create this vocabulary we need to pad our sentences (just like for counting ngrams) and then combine the sentences into one flat stream of words.


To do the above steps, we need to create an Ngram and add the padding then flatten the list. All these steps are requested for pre-processing to train the MLE model. However, we can use padded_everygram_pipeline() in NLTK to do all the steps in one call.

The  padded_everygram_pipeline() will return:  
1- **ngrams_gen:** Each sentence becomes a list of n-grams with padding.  
2- **vocab_gen:** A flat stream of all tokens, with start and end symbols included  

**References:**

https://www.nltk.org/api/nltk.lm.html

https://www.kaggle.com/code/alvations/n-gram-language-model-with-nltk

```python
from nltk.lm.preprocessing import padded_everygram_pipeline
from nltk.util import ngrams
from nltk.tokenize import word_tokenize
from nltk.lm import smoothing

string = "moses supposes his toeses are roses but moses supposes erroneously"

#Create Bigram Language Model
train, vocab = padded_everygram_pipeline(2, [word_tokenize(string)])
```

```python
from nltk.lm import MLE
lm = MLE(2)
lm.fit(train, vocab) #Train the Maximum Likelihood Model
```

### Calculating Probabilities
Here’s how you get the score for a word given some preceding context. For example we want to know what is the chance that “b” is preceded by “a”.

```python
# P(supposes | moses)
print(lm.score("supposes", ["moses"]))
# P(moses | supposes): this bigram is unseen in the example.
print(lm.score("moses", ["supposes"]))
```

### Evaluating n_grams models using preplexity

**Preplexity** is a measurement of how well a probability model (like a language model) predicts a sample.

It's the probability of the test set normalized by the number of words.

A `lower` perplexity means:

- The model is less “perplexed” by the data.

- The model assigns higher probabilities to the actual words in the test set.

- Therefore, better performance.

A `higher perplexity` = more uncertainty = worse model.

```python
test = [('moses', 'supposes'), ('toeses','are')]
print(lm.perplexity(test)) #Perplexity for bigram model
```

### 1. Load the Data

```python
from pathlib import Path
import re
import math
import emoji
import kagglehub
import pandas as pd
from sklearn.model_selection import train_test_split
from nltk.tokenize import word_tokenize
from nltk.lm import MLE
from nltk.lm.preprocessing import padded_everygram_pipeline, pad_both_ends
from nltk.util import ngrams

local_folder = Path("pakistan_tweets_data")
if local_folder.is_dir() and any(local_folder.rglob("*.csv")):
    dataset_folder = local_folder
else:
    dataset_folder = Path(kagglehub.dataset_download(
        "adizafar/large-random-tweets-from-pakistan"
    ))

csv_files = sorted(dataset_folder.rglob("*.csv"))
print("CSV files:", [str(p.relative_to(dataset_folder)) for p in csv_files])
csv_filename = None  # Set a relative filename if there is more than one CSV.
if csv_filename is not None:
    csv_path = dataset_folder / csv_filename
elif len(csv_files) == 1:
    csv_path = csv_files[0]
else:
    raise ValueError("Check the file list and set csv_filename to the tweet CSV.")

try:
    data = pd.read_csv(csv_path, encoding="utf-8")
except UnicodeDecodeError:
    data = pd.read_csv(csv_path, encoding="cp1252")

print("Dataset shape:", data.shape)
print("Columns:", data.columns.tolist())
data.head()
```

### Select the Tweet Text Column
Set `text_column` manually if more than one candidate column is present. A selected column must contain the actual tweet text.

```python
text_column = None
if text_column is None:
    candidates = [column for column in data.columns if column.strip().lower() in
                  {"text", "tweet", "tweets", "tweet_text", "full_text", "content"}]
    if len(candidates) != 1:
        raise ValueError("Set text_column to the column containing tweet text.")
    text_column = candidates[0]
if text_column not in data.columns:
    raise ValueError(f"Column not found: {text_column}")
print("Tweet column:", text_column)
```

### 2. Pre-process the Tweets
Remove websites, hashtag tokens, mentions, the standalone `RT` marker, and emojis. Then lowercase the remaining text and normalize whitespace. This preserves word order for the bigram model. Hashtag words are removed with their `#` markers as requested.

```python
def clean_tweet(text):
    text = re.sub(r"https?://\S+|www\.\S+", " ", text, flags=re.IGNORECASE)
    text = re.sub(r"(?<![\w#])#\w+", " ", text)
    text = re.sub(r"(?<!\w)@\w+", " ", text)
    text = re.sub(r"\bRT\b", " ", text, flags=re.IGNORECASE)
    text = emoji.replace_emoji(text, replace=" ")
    return re.sub(r"\s+", " ", text.lower()).strip()

cleaned_tweets = data[text_column].dropna().astype(str).apply(clean_tweet)
cleaned_tweets = cleaned_tweets[cleaned_tweets.str.len() > 0].drop_duplicates()

print(f"Nonempty, unique cleaned tweets: {len(cleaned_tweets):,}")
cleaned_tweets.head()
```

### 3. Build the Bigram MLE Model
Split the cleaned tweets into 80% training data and 20% held-out test data using a fixed random seed. Duplicate cleaned tweets are removed before splitting to avoid identical tweets appearing in both sets. Fit the vocabulary and bigram counts using the training tweets only. Boundary tokens mark the beginning and end of each tweet.

```python
if len(cleaned_tweets) < 5:
    raise ValueError("At least five nonempty, unique tweets are needed for a train/test split.")

train_tweets, test_tweets = train_test_split(
    cleaned_tweets.tolist(), test_size=0.2, random_state=42
)
train_tokens = [word_tokenize(tweet) for tweet in train_tweets]
test_tokens = [word_tokenize(tweet) for tweet in test_tweets]

train_ngrams, vocabulary = padded_everygram_pipeline(2, train_tokens)
tweet_model = MLE(2)
tweet_model.fit(train_ngrams, vocabulary)

print(f"Training tweets: {len(train_tokens):,}")
print(f"Test tweets: {len(test_tokens):,}")
print(f"Vocabulary size: {len(tweet_model.vocab):,}")
```

### 4. Evaluate the Model
Calculate perplexity on held-out bigrams. Lower perplexity indicates better predictions. An unsmoothed MLE model assigns zero probability to unseen bigrams, so its test perplexity can be infinite. Report that result and the number of zero-probability bigrams rather than hiding them.

```python
test_bigrams = [
    bigram
    for tokens in test_tokens
    for bigram in ngrams(pad_both_ends(tokens, n=2), 2)
]
if not test_bigrams:
    raise ValueError("No test bigrams were produced.")

zero_probability_count = sum(
    tweet_model.score(word, [previous_word]) == 0
    for previous_word, word in test_bigrams
)
test_perplexity = tweet_model.perplexity(test_bigrams)

print(f"Held-out bigrams: {len(test_bigrams):,}")
print(f"Zero-probability bigrams: {zero_probability_count:,}")
print("Test perplexity:", test_perplexity)
if math.isinf(test_perplexity):
    print("Infinite perplexity: the unsmoothed MLE model encountered unseen bigrams.")
```

### 5. Probability of the Bigram “pakistan is”
The requested conditional probability is `P(is | pakistan)`, calculated from the training model. If the context or bigram was not observed, MLE returns zero.

```python
pakistan_is_probability = tweet_model.score("is", ["pakistan"])
print("P(is | pakistan):", pakistan_is_probability)
print("Training count of (pakistan, is):", tweet_model.counts[["pakistan"]]["is"])
```

### 6. Perplexity of “pakistan”
A single word is a unigram, so evaluating `[('pakistan',)]` uses the model's unigram probability. For a nonzero probability, its perplexity is `1 / P(pakistan)`. To distinguish this from the bigram interpretation, also evaluate the one-word sentence using `(<s>, pakistan)` and `(pakistan, </s>)`. Either result can be infinite if the corresponding event is unseen.

```python
pakistan_probability = tweet_model.score("pakistan")
pakistan_perplexity = tweet_model.perplexity([("pakistan",)])
pakistan_sentence_bigrams = list(ngrams(pad_both_ends(["pakistan"], n=2), 2))
pakistan_sentence_perplexity = tweet_model.perplexity(pakistan_sentence_bigrams)

print("P(pakistan):", pakistan_probability)
print("Unigram perplexity of pakistan:", pakistan_perplexity)
print("One-word sentence bigram perplexity:", pakistan_sentence_perplexity)
```

### Generate Tweets
Start with `<s>`, sample the next word from the bigram model, and stop at `</s>` or the word limit. Fixed seeds make the examples reproducible. This small language model can repeat training phrases; generated text is not a claim about real events.

```python
def generate_tweet(model, max_words=20, seed=42):
    context = ["<s>"]
    generated_words = []
    for step in range(max_words):
        word = model.generate(1, text_seed=context, random_seed=seed + step)
        if word == "</s>":
            break
        if word not in {"<s>", "<UNK>"}:
            generated_words.append(word)
        context = [word]
    return " ".join(generated_words)

for number, seed in enumerate([42, 43, 44], start=1):
    print(f"Generated tweet {number}:", generate_tweet(tweet_model, seed=seed))
```
