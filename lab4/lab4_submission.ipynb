# Disneyland Reviews Dataset
In this lab we'll look into Disneyland Reviews dataset.  
*Link: https://www.kaggle.com/datasets/arushchillar/disneyland-reviews*

The dataset includes 42,000 reviews of 3 Disneyland branches - Paris, California and Hong Kong, posted by visitors on Trip Advisor.

Column Description:

- Review_ID: unique id given to each review
- Rating: ranging from 1 (unsatisfied) to 5 (satisfied)
- Year_Month: when the reviewer visited the theme park
- Reviewer_Location: country of origin of visitor
- Review_Text: comments made by visitor
- Disneyland_Branch: location of Disneyland Park

```python
from pathlib import Path
import kagglehub
import pandas as pd

local_csv = Path("DisneylandReviews.csv")
if local_csv.is_file():
    disneyland_csv = local_csv
else:
    disneyland_folder = Path(kagglehub.dataset_download("arushchillar/disneyland-reviews"))
    matches = list(disneyland_folder.rglob("DisneylandReviews.csv"))
    if len(matches) != 1:
        raise FileNotFoundError("Place DisneylandReviews.csv beside this notebook.")
    disneyland_csv = matches[0]

data = pd.read_csv(disneyland_csv, encoding="latin-1")
data.head()
```

```python
#Check for missing values
data.isna().sum()
```

```python
data = data.dropna(subset=['Review_Text', 'Rating']).copy()
```

```python
# Map star ratings to sentiments
data['Sentiment'] = data['Rating'].apply(lambda rating: 'positive' if rating > 3 else ('negative' if rating < 3 else 'neutral'))
data.head()
```

```python
# Import necessary libraries
import re
import nltk

def preprocess_text(text):

    # 1. Convert text to lowercase
    text = text.lower()

    # 2. Remove punctuation and special characters
    text = re.sub(r'[^a-z\s]', '', text)

    # 3. Tokenize the text
    words = text.split()

    return ' '.join(words)

# Apply preprocessing to the review text
data['Review_Text'] = data['Review_Text'].apply(preprocess_text)

# Display cleaned reviews
data[['Review_Text', 'Sentiment']].head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

train_data, test_data, train_labels, test_labels = train_test_split(
    data["Review_Text"], data["Sentiment"], test_size=0.2,
    random_state=42, stratify=data["Sentiment"]
)
tfidf_vectorizer = TfidfVectorizer(max_features=1000)
train_vectors = tfidf_vectorizer.fit_transform(train_data)
test_vectors = tfidf_vectorizer.transform(test_data)
```

```python
print(f"train_data size: {train_data.shape}")
print(f"test_data size: {test_data.shape}")
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Train the Support Vector Machine (SVM) classifier
svm_classifier = SVC(kernel='linear')
svm_classifier.fit(train_vectors, train_labels)

# Predictions on the test set
predictions = svm_classifier.predict(test_vectors)

# Evaluate the model
accuracy = accuracy_score(test_labels, predictions)
print(f"Accuracy: {accuracy:.2f}")

# Display classification report
print("Classification Report:")
print(classification_report(test_labels, predictions, zero_division=0))
```

### Task 1: Load the Data and Apply Five Pre-processing Steps
If downloading manually, extract the Amazon CSV into `amazon_reviews_data` beside this notebook. The expected review and star-rating columns are `Reviews` and `Rating`. Ratings 1–2 are negative, 3 is neutral, and 4–5 are positive.

```python
from pathlib import Path
import re
import kagglehub
import pandas as pd
from nltk.corpus import stopwords

local_folder = Path("amazon_reviews_data")
if local_folder.is_dir() and any(local_folder.rglob("*.csv")):
    amazon_folder = local_folder
else:
    amazon_folder = Path(kagglehub.dataset_download(
        "promptcloudhq/amazon-reviews-unlocked-mobile-phones"
    ))

csv_files = sorted(amazon_folder.rglob("*.csv"))
print("CSV files:", [str(p.relative_to(amazon_folder)) for p in csv_files])
csv_filename = None  # Set a relative filename here if there is more than one CSV.
if csv_filename is not None:
    amazon_csv = amazon_folder / csv_filename
elif len(csv_files) == 1:
    amazon_csv = csv_files[0]
else:
    raise ValueError("Check the file list and set csv_filename to the reviews CSV.")

try:
    reviews = pd.read_csv(amazon_csv, encoding="utf-8")
except UnicodeDecodeError:
    reviews = pd.read_csv(amazon_csv, encoding="cp1252")

required_columns = {"Reviews", "Rating"}
if not required_columns.issubset(reviews.columns):
    raise ValueError(f"Expected columns {required_columns}; found {reviews.columns.tolist()}")

print("Dataset shape:", reviews.shape)
reviews[["Reviews", "Rating"]].head()
```

#### Check Missing Values and Prepare Sentiment Labels
Remove rows without usable review text or a valid 1–5 star rating. Preserve the original review text for inspection.

```python
print("Missing values before cleaning:")
print(reviews[["Reviews", "Rating"]].isna().sum())

reviews = reviews.dropna(subset=["Reviews", "Rating"]).copy()
reviews["Rating"] = pd.to_numeric(reviews["Rating"], errors="coerce")
reviews = reviews[reviews["Rating"].isin([1, 2, 3, 4, 5])].copy()
reviews["Reviews"] = reviews["Reviews"].astype(str)
reviews["Sentiment"] = reviews["Rating"].map({
    1: "negative", 2: "negative", 3: "neutral", 4: "positive", 5: "positive"
})
print("Sentiment counts before text preprocessing:")
print(reviews["Sentiment"].value_counts())
```

#### Five Text Pre-processing Steps
1. Convert text to lowercase.
2. Remove URLs and HTML tags.
3. Replace punctuation, digits, and other non-letter characters with spaces.
4. Tokenize the cleaned text into words.
5. Remove English stop words while retaining negation (`no`, `not`, `nor`, `never`).

Expand `n't` to `not` before removing punctuation so contractions such as `isn't` keep their negative meaning. Rejoin the tokens for TF-IDF input.

```python
english_stop_words = set(stopwords.words("english")) - {"no", "not", "nor", "never"}

def preprocess_review(text):
    # 1. Lowercase.
    text = text.lower()
    text = text.replace("’", "'")
    text = re.sub(r"n['’]t\b", " not", text)

    # 2. Remove URLs and HTML tags.
    text = re.sub(r"https?://\S+|www\.\S+|<[^>]+>", " ", text)

    # 3. Remove punctuation, digits, and special characters.
    text = re.sub(r"[^a-z\s]", " ", text)

    # 4. Tokenize.
    tokens = text.split()

    # 5. Remove stop words but keep negation.
    tokens = [token for token in tokens if token not in english_stop_words]
    return " ".join(tokens)

reviews["Clean_Review"] = reviews["Reviews"].apply(preprocess_review)
reviews = reviews[reviews["Clean_Review"].str.len() > 0].copy()
reviews[["Reviews", "Clean_Review", "Sentiment"]].head()
```

#### Avoid Duplicate Text Across Training and Test Data
Group identical cleaned reviews. Remove groups with conflicting labels and retain one copy of each remaining review, preventing identical cleaned text from appearing in both sets.

```python
before_deduplication = len(reviews)
labels_per_text = reviews.groupby("Clean_Review")["Sentiment"].transform("nunique")
conflicting_rows = int((labels_per_text > 1).sum())
reviews = reviews[labels_per_text == 1].drop_duplicates(subset=["Clean_Review"]).copy()

print("Rows removed with conflicting labels:", conflicting_rows)
print("Total rows removed during deduplication:", before_deduplication - len(reviews))
print("Reviews available for modelling:", len(reviews))
print(reviews["Sentiment"].value_counts())
```

### Task 2: Split Training and Test Data
Use an 80/20 stratified split so each sentiment class is represented. A fixed random seed makes the split reproducible. The test set stays separate from fitting the vectorizer and classifier.

```python
from sklearn.model_selection import train_test_split

class_counts = reviews["Sentiment"].value_counts()
if set(class_counts.index) != {"negative", "neutral", "positive"} or class_counts.min() < 2 or len(reviews) < 11:
    raise ValueError("Provide at least 11 unique reviews overall and at least two in each sentiment class.")

X_train, X_test, y_train, y_test = train_test_split(
    reviews["Clean_Review"], reviews["Sentiment"],
    test_size=0.2, random_state=42, stratify=reviews["Sentiment"]
)

print("Training reviews:", len(X_train))
print("Test reviews:", len(X_test))
print("Training class counts:")
print(y_train.value_counts())
print("Test class counts:")
print(y_test.value_counts())
```

### Task 3: Extract TF-IDF Features
Fit the vectorizer on training text only, then use the learned vocabulary to transform the test text. Keep sparse matrices to avoid unnecessary memory use.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=10000, ngram_range=(1, 2))
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Training feature matrix:", X_train_tfidf.shape)
print("Test feature matrix:", X_test_tfidf.shape)
```

### Task 4: Train a Naive Bayes Classifier
Multinomial Naive Bayes accepts nonnegative text features such as TF-IDF. Train using only the training features and labels.

```python
from sklearn.naive_bayes import MultinomialNB

classifier = MultinomialNB(alpha=1.0)
classifier.fit(X_train_tfidf, y_train)
print("Naive Bayes training completed.")
```

### Task 5: Evaluate on the Test Set
Report accuracy and precision, recall, and F1 for all three classes. Macro F1 weights each class equally, which helps interpret performance when the dataset is imbalanced. Report computed values without assuming the model will perform well.

```python
from sklearn.metrics import accuracy_score, classification_report, f1_score

predictions = classifier.predict(X_test_tfidf)
label_order = ["negative", "neutral", "positive"]

accuracy = accuracy_score(y_test, predictions)
macro_f1 = f1_score(y_test, predictions, labels=label_order, average="macro", zero_division=0)
print(f"Test accuracy: {accuracy:.4f}")
print(f"Macro F1: {macro_f1:.4f}")
print("Classification report:")
print(classification_report(y_test, predictions, labels=label_order, zero_division=0))
```

### Task 6: Print the Confusion Matrix
Rows show actual sentiment and columns show predicted sentiment. Diagonal entries are correct predictions; off-diagonal entries show misclassifications.

```python
from sklearn.metrics import confusion_matrix

matrix = confusion_matrix(y_test, predictions, labels=label_order)
confusion_table = pd.DataFrame(
    matrix,
    index=pd.Index(label_order, name="Actual"),
    columns=pd.Index(label_order, name="Predicted"),
)
print("Confusion matrix:")
print(confusion_table.to_string())
confusion_table
```
